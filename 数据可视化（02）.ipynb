```python
%matplotlib inline

import pandas as pd
import matplotlib.pyplot as plt

# ============================
# 1. 设置中文字体
# ============================
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# ============================
# 2. 输入你的数据
# ============================
df = pd.DataFrame({
    '区域': ['华北', '华南', '东北', '西北', '西南', '华东'],
    '销售量': [2354, 1902, 3524, 2698, 2896, 2563]
})

# ============================
# 3. 计算平均值
# ============================
average = df['销售量'].mean()

# 最高销售区域
max_row = df.loc[df['销售量'].idxmax()]
max_region = max_row['区域']
max_sales = max_row['销售量']

# 最低销售区域
min_row = df.loc[df['销售量'].idxmin()]
min_region = min_row['区域']

# 最高区域占总销量比例
total_sales = df['销售量'].sum()
max_percentage = max_sales / total_sales * 100

# ============================
# 4. 创建画布
# ============================
fig, ax = plt.subplots(figsize=(12, 7))

# 深蓝色背景
bg_color = '#171d47'

fig.patch.set_facecolor(bg_color)
ax.set_facecolor(bg_color)

# ============================
# 5. 绘制柱状图
# ============================
bars = ax.bar(
    df['区域'],
    df['销售量'],
    width=0.32,
    color='#087fc1'
)

# ============================
# 6. 设置Y轴
# ============================
ax.set_ylim(0, 4000)

ax.set_yticks([0, 1000, 2000, 3000, 4000])

ax.tick_params(
    axis='y',
    colors='white',
    labelsize=11,
    length=0
)

ax.tick_params(
    axis='x',
    colors='white',
    labelsize=12,
    length=0,
    pad=10
)

# ============================
# 7. 添加横向虚线
# ============================
ax.grid(
    axis='y',
    linestyle='--',
    linewidth=1,
    color='white',
    alpha=0.20
)

ax.set_axisbelow(True)

# 去掉边框
for spine in ax.spines.values():
    spine.set_visible(False)

# ============================
# 8. 柱子上显示销售量
# ============================
for bar, value in zip(bars, df['销售量']):
    
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 80,
        str(value),
        ha='center',
        va='bottom',
        color='white',
        fontsize=10
    )

# ============================
# 9. 绘制平均值黄色水平线
# ============================
ax.axhline(
    y=average,
    color='#ffc000',
    linewidth=1.5
)

# ============================
# 10. 平均值标签
# ============================
ax.text(
    len(df) - 0.15,
    average + 35,
    f'平均值：{average:.0f}',
    color='#ffc000',
    fontsize=10,
    ha='right',
    va='bottom'
)

# ============================
# 11. 标题
# ============================
fig.text(
    0.08,
    0.90,
    '3月各区域销量分布',
    fontsize=24,
    fontweight='bold',
    color='white'
)

# ============================
# 12. 副标题
# ============================
subtitle = (
    f'{max_region}销量最多占比总销量的{max_percentage:.0f}%，'
    f'{min_region}销量最低'
)

fig.text(
    0.08,
    0.83,
    subtitle,
    fontsize=15,
    color='white'
)

# ============================
# 13. 底部说明
# ============================
fig.text(
    0.065,
    0.08,
    '注：数据来源于公司销售系统，统计日期截至2022.03.31',
    fontsize=9,
    color='white'
)

# ============================
# 14. 调整布局
# ============================
plt.subplots_adjust(
    left=0.13,
    right=0.95,
    bottom=0.18,
    top=0.68
)

# ============================
# 15. 显示图片
# ============================
plt.show()
```
