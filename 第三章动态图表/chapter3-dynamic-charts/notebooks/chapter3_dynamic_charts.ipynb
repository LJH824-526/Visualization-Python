## 使用

在项目根目录安装依赖：`python -m pip install -r requirements-notebook.txt`。打开本文件，选择 Python 3 内核，先执行公共设置，再执行需要的图表单元格。

已保存的 PNG 是默认状态预览；**重新执行单元格后才会出现实时控件**。若提示缺少 ipywidgets，请安装 Notebook 依赖并重启内核。也可离线打开根目录 `interactive.html`，体验预生成的 42 种选择状态。

原始宏不需要运行。人力资源数据仅保留部门、学历、人数和收入合计，用于精确重算均值，未包含员工号或逐人明细。字体、局部半径与标签排布可能与 Excel 渲染略有差异，不能保证逐像素相同。
